# Integrals calculation

```python
num_points = 300

def mapping_func(x):
    return ((1 + x) / (1 - x)) ** 3

def s_vector(n):
    return np.cos(np.pi / n * (np.arange(1, n + 1) - 0.5))

def z_vector(n):
    return np.cos(np.pi / n * np.arange(1, n))
    
xi_p = mapping_func(s_vector(num_points))[::-1]
xi_c = mapping_func(z_vector(num_points))[::-1]

power_1 = 1/2 
power_2 = 1/3

beta = 0.25
```

```python
p_s_near_field_integral = integral_near_tip_log(xi_c, xi_p[0], power_1, p_s)
p_s_matrix_0 = integral_intervals_log(xi_c, xi_p, power_1, p_s)
p_s_matrix_inf = integral_intervals_log(xi_c, xi_p, power_2, p_s)
```

```python
p_e_kernel = lambda x: p_e(x, False)

p_e_near_field_integral = integral_near_tip_cauchy(xi_c, xi_p[0], power_1, p_e_kernel)
p_e_matrix_0 = integral_intervals_cauchy(xi_c, xi_p, power_1, p_e_kernel)
p_e_matrix_inf = integral_intervals_cauchy(xi_c, xi_p, power_2, p_e_kernel)
```

```python
s_e_kernel = lambda x, beta=beta: sigma_yy_e(x, beta, False)

s_e_near_field_integral = integral_near_tip_cauchy(xi_c, xi_p[0], power_1, s_e_kernel)
s_e_matrix_0 = integral_intervals_cauchy(xi_c, xi_p, power_1, s_e_kernel)
s_e_matrix_inf = integral_intervals_cauchy(xi_c, xi_p, power_2, s_e_kernel)
```

```python
s_e_kernel = lambda x: sigma_yy_e(x, 0, False)

cauchy_near_field_integral = integral_near_tip_cauchy(xi_c, xi_p[0], power_1, s_e_kernel)
cauchy_matrix_0 = integral_intervals_cauchy(xi_c, xi_p, power_1, s_e_kernel)
cauchy_matrix_inf = integral_intervals_cauchy(xi_c, xi_p, power_2, s_e_kernel)
```
